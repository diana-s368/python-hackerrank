Given the names and grades for each student in a class of n students, store them in a nested list and print the name(s) of any student(s) having the second lowest grade.

### Task 4: Finding the percentage

The provided code stub will read in a dictionary containing key/value pairs of name:[marks] for a list of students. Print the average of the marks array for the student name provided, showing 2 places after the decimal.

3 \
Krishna 67 68 69 \
Arjun 70 98 63 \

```python
data = """3
Krishna 67 68 69
Arjun 70 98 63
Malika 52 56 60
Malika"""

lines = data.splitlines()

n = int(lines[0])
student_marks = {}

for i in range(1, n + 1):
    name, *scores = lines[i].split()
    student_marks[name] = [float(score) for score in scores]

query_name = lines[n + 1]

avg = sum(student_marks[query_name]) / len(student_marks[query_name])
print(f"{avg:.2f}")
```

### Task 7: Text Alignment

You are given a partial code that is used for generating the HackerRank Logo of variable thickness.
Your task is to replace the blank (______) with rjust, ljust or center.

```python
# read thickness
t = int(input().strip())
c = 'H'

# Top Cone
for i in range(t):
    print((c*i).rjust(t-1) + c + (c*i).ljust(t-1))

# Top Pillars
for _ in range(t + 1):
    print((c*t).center(2*t) + (c*t).center(6*t))

# Middle Belt
for _ in range((t + 1) // 2):
    print((c*t*5).center(6*t))

# Bottom Pillars
for _ in range(t + 1):
    print((c*t).center(2*t) + (c*t).center(6*t))

# Bottom Cone
for i in range(t):
    print(((c*(t - i - 1)).rjust(t) + c + (c*(t - i - 1)).ljust(t)).rjust(6*t))
```

### Task 10: Alhpabet Rangoli

You are given an integer, . Your task is to print an alphabet rangoli of size . (Rangoli is a form of Indian folk art based on creation of patterns.)

```python
import string

n = int(input().strip())
letters = string.ascii_lowercase

rows = []

for i in range(n):
    left = letters[n-1:i:-1]
    right = letters[i:n]
    row = '-'.join(left + right)
    rows.append(row.center(4*n-3, '-'))

for row in rows[::-1]:
    print(row)

for row in rows[1:]:
    print(row)
```

### Task 1: Collection counter()

Raghu is a shoe shop owner. His shop has N number of shoes.
He has a list containing the size of each shoe he has in his shop.
There are N number of customers who are willing to pay x_i amount of money only if they get the shoe of their desired size.

Your task is to compute how much money  earned.

```python
from collections import Counter
```

```python
X = int(input())
shoe_sizes = Counter(map(int, input().split()))

N = int(input())
total = 0
```

```python

for _ in range(N):
    size, price = map(int, input().split())

    if shoe_sizes[size] > 0:
        total += price
        shoe_sizes[size] -= 1

print(total)
```

### Task 2: defaultdict 



```python
from collections import defaultdict
```

```python

n, m = map(int, input().split())

A = defaultdict(list)

for i in range(1, n + 1):
    word = input()
    A[word].append(i)

for _ in range(m):
    word = input()

    if word in A:
        print(*A[word])
    else:
        print(-1)
```

## Map and Lambda Function


```python
cube = lambda x: x ** 3

def fibonacci(n):
    fib = []
    a, b = 0, 1

    for _ in range(n):
        fib.append(a)
        a, b = b, a + b

    return fib

if __name__ == '__main__':
    n = int(input())
    print(list(map(cube, fibonacci(n))))
```
